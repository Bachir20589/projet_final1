```python
import pandas as p
import matplotlib.pyplot as mp
import seaborn as sn

with open("etudiants.csv") as f:
    a=p.read_csv(f)
    b=a.isna().sum()
    a.fillna({"age":21},inplace=True)
    a.dropna(inplace=True)
    a.fillna({"filiere":"Informatique"},inplace=True)
    for i in a.index:
        if a.loc[i,"note"]=="NaN":
            a.drop(i,inplace=True)
    c=a["note"].mean()

    for i in a.index:
        if a.loc[i,"note"]>=14:
            print(a.loc[i,"nom"])
    a.drop_duplicates(inplace=True)
    a["filiere"].value_counts().plot(kind="bar")
    mp.xlabel("Nombre Etudiant")
    mp.ylabel("Filiere")
    mp.title("Nombre d'étudiant par filière")
    mp.show()
    mp.ylabel("age")
    mp.xlabel("note")
    mp.show()
    sn.heatmap(a.corr(numeric_only=True),annot=True,cmap="coolwarm")
    mp.show()
    sn.histplot(data=a,x="note",bins=10,edgecolor="red")
    mp.show()
    sn.scatterplot(data=a,x="age",y="note")
    mp.show()
    d=a.corr(numeric_only=True)
print(a)
print(b)
print(a.head(5))
print(c)
print(d)
a.to_csv("etudiant.csv",index=False)
```
